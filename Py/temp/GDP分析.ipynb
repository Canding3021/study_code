```python
import pandas as pd
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif'] = ['Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False
df = pd.read_csv('data/GDP.csv')
df.columns = ['quarter', '第一产业GDP', '第二产业GDP', '第三产业GDP']
df = df.sort_values('quarter').reset_index(drop=True)
df
```

```python
mean_primary = df['第一产业GDP'].mean()
print(f'第一产业GDP平均值:{mean_primary:.2f} 亿元')
```

```python
fig, ax = plt.subplots(figsize=(10, 5))
ax.bar(df['quarter'].astype(str), df['第一产业GDP'],color= '#438EDB', width=0.62, zorder=3)
ax.set_title('第一产业GDP季度数据', fontsize=15, fontweight='bold', pad=14)
ax.set_xlabel('季度')
ax.set_ylabel('GDP(亿元)')
ax.yaxis.set_major_formatter(lambda x, pos: f'{x / 10000:.1f}万')
ax.grid(axis='y', linestyle='--', alpha=0.4, zorder=0)
ax.set_axisbelow(True)
for spine in ('top', 'right'):
    ax.spines[spine].set_visible(False)
ax.tick_params(axis='x', rotation=45)
plt.tight_layout()
plt.show()
```
