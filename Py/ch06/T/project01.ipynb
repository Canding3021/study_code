```python
import os
from PIL import Image
import numpy as np
```

```python
def img2array(imgfile, size=(32, 32)):
    img = Image.open(imgfile)
    img = img.resize(size, Image.LANCZOS)
    img = img.convert('L')
    np_img = np.asarray(img)
    vect = np_img.reshape(-1)  # 自动计算后一个维度
    return vect
```

```python
dataOne = img2array('data/cat.jpg')
dataOne.shape
```

```python
def convert2dataset(file_path, size=(32, 32)):
    list_file = os.listdir(file_path)
    file_num = len(list_file)
    datas = np.zeros((file_num, size[0]*size[1]))
    labels = []
    for i in range(file_num):
        animal_name = list_file[i].split('_')[0]
        label = 0 if animal_name=='cat' else 1
        labels.append(label)
        datas[i, :] = img2array(os.path.join(file_path, list_file[i]), size=size)
    return datas, labels
```

```python
x_train, y_train = convert2dataset('data/train')
x_test, y_test = convert2dataset('data/test')
```
